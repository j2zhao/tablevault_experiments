```python
def get_embeddings(text):
    return client.embeddings.create(
            input=text,
            model="text-embedding-3-large"
        ).data[0].embedding
```

```python
import torch
import numpy as np
from datasets import Dataset
from sentence_transformers import SentenceTransformer
from sklearn.metrics import accuracy_score, f1_score, classification_report
from tqdm.auto import tqdm

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
print("device:", device)
```

```python
model_name = "sentence-transformers/all-MiniLM-L6-v2"
model = SentenceTransformer(model_name, device=str(device))
model.max_seq_length = 64

print("model:", model_name)
print("max_seq_length:", model.max_seq_length)
```

```python
ds = vault.query_item_content("glue_mrpc_validation")
ds = Dataset.from_list(ds)
print(ds)
print(ds[0])

sent1 = ds["sentence1"]
sent2 = ds["sentence2"]
y_true = np.array(ds["label"])

print("num_examples:", len(y_true))
print("positive_rate:", float(y_true.mean()))
```

```python
batch_size = 128

def batched_encode(texts, desc):
    batches = []
    for i in tqdm(range(0, len(texts), batch_size), desc=desc):
        batch = texts[i:i + batch_size]
        emb = model.encode(
            batch,
            batch_size=batch_size,
            convert_to_tensor=True,
            normalize_embeddings=True,
            show_progress_bar=False,
        )
        batches.append(emb)
    return torch.cat(batches, dim=0)

emb1 = batched_encode(sent1, "Encoding sentence1")
emb2 = batched_encode(sent2, "Encoding sentence2")

print("emb1 shape:", tuple(emb1.shape))
print("emb2 shape:", tuple(emb2.shape))
```

```python
emb1_list = emb1.cpu().float().tolist()
emb2_list = emb2.cpu().float().tolist()
vault.create_embedding_list("sentence-transformers-sentence-1-max64", ndim=384)
vault.create_embedding_list("sentence-transformers-sentence-2-max64", ndim=384)

for i in range(len(emb1_list)):
    vault.append_embedding("sentence-transformers-sentence-1-max64", emb1_list[i], 
                       input_items = {"glue_mrpc_validation": [i, i + 1]}
                       )
    vault.append_embedding("sentence-transformers-sentence-2-max64", emb2_list[i], 
                       input_items = {"glue_mrpc_validation": [i, i + 1]}
                       )


description = "Embedding list containing one 384-dimensional normalized sentence embedding per row for the sentence1 field of the GLUE MRPC validation dataset. Each embedding is generated with the sentence-transformers/all-MiniLM-L6-v2 model using max_seq_length=64, so inputs may be truncated beyond 64 tokens. Structure: this dataset stores a single embedding vector for each original example and is linked back to the corresponding row in glue_mrpc_validation via input item indices. In this workflow, it provides the vector representation of the first sentence in each pair, which is later compared with the matching sentence-transformers-sentence-2-max64 embedding using cosine similarity to produce paraphrase predictions."
embedding = get_embeddings(description)
vault.create_description("sentence-transformers-sentence-1-max64", description, embedding)

properties = {"task": "paraphrase detection", "representation": "sentence embedding", "embedding_model": "sentence-transformers/all-MiniLM-L6-v2", "embedding_dim": "384", "normalized": "true", "source": "glue/mrpc", "split": "validation", "source_field": "sentence1", "size": "408", "max_seq_length": "64", "similarity_metric": "cosine", "framework": "sentence-transformers", "language": "english"} #e.g. task: paraphrase detection

for prop, cat in properties.items():
    embedding = get_embeddings(prop)
    vault.create_description("sentence-transformers-sentence-1-max64", cat, embedding, prop)

description = "Embedding dataset containing one 384-dimensional normalized sentence embedding per row for the sentence2 field of each example in the GLUE MRPC validation set. Each vector was produced with the sentence-transformers/all-MiniLM-L6-v2 model using a maximum sequence length of 64, so it represents the truncated second sentence in a paraphrase pair. The dataset structure is an embedding list: each item stores a single 384-float embedding and is linked back to the corresponding row in glue_mrpc_validation. In this workflow, these sentence2 embeddings are paired with the matching sentence-transformers-sentence-1-max64 embeddings to compute cosine similarity scores and generate paraphrase predictions."
embedding = get_embeddings(description)
vault.create_description("sentence-transformers-sentence-2-max64", description, embedding)

properties = {"type": "embedding", "modality": "text", "task": "paraphrase detection", "source": "glue/mrpc", "dataset": "MRPC", "split": "validation", "text_field": "sentence2", "model": "sentence-transformers/all-MiniLM-L6-v2", "embedding_dim": "384", "max_seq_length": "64", "normalized": "True", "size": "408", "domain": "news"} #e.g. task: paraphrase detection

for prop, cat in properties.items():
    embedding = get_embeddings(prop)
    vault.create_description("sentence-transformers-sentence-2-max64", cat, embedding, prop)
```

```python
cosine_scores = (emb1 * emb2).sum(dim=1)
threshold = 0.80
y_pred = (cosine_scores >= threshold).to(torch.int64).cpu().numpy()
scores_np = cosine_scores.cpu().numpy()


vault.create_record_list("sentence_transformers_mrpc_prediction_max64", column_names=["prediction"])

for i in range(len(y_pred)):
    vault.append_record("sentence_transformers_mrpc_prediction_max64", {"prediction": int(y_pred[i])}, 
                       input_items = {
                           "glue_mrpc_validation": [i, i + 1],
                           "sentence-transformers-sentence-1-max64": [i, i + 1],
                           "sentence-transformers-sentence-2-max64": [i, i + 1],
                       }
                       )

description = "Binary paraphrase predictions for the GLUE MRPC validation set, generated by encoding each sentence pair with sentence-transformers/all-MiniLM-L6-v2 using max_seq_length=64, computing the cosine similarity between the normalized sentence1 and sentence2 embeddings, and applying a threshold of 0.80. This record list contains one row per validation example with a single field, prediction, where 1 indicates predicted paraphrase and 0 indicates predicted non-paraphrase. Each record is linked back to the corresponding source example in glue_mrpc_validation and to the two embedding lists sentence-transformers-sentence-1-max64 and sentence-transformers-sentence-2-max64. In this workflow, the dataset serves as the stored model output used for downstream evaluation, error analysis, and experiment summary reporting."
embedding = get_embeddings(description)
vault.create_description("sentence_transformers_mrpc_prediction_max64", description, embedding)

properties = {"task": "paraphrase detection", "output_type": "binary prediction", "source": "glue/mrpc", "split": "validation", "size": "408", "domain": "newswire", "model": "sentence-transformers/all-MiniLM-L6-v2", "input_type": "sentence pair", "embedding_dim": "384", "max_seq_length": "64", "similarity_metric": "cosine", "threshold": "0.80"} #e.g. task: paraphrase detection

for prop, cat in properties.items():
    embedding = get_embeddings(prop)
    vault.create_description("sentence_transformers_mrpc_prediction_max64", cat, embedding, prop)




acc = accuracy_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
report = classification_report(y_true, y_pred, target_names=["not_paraphrase", "paraphrase"])

print({"accuracy": acc, "f1": f1, "threshold": threshold})
print(classification_report(y_true, y_pred, target_names=["not_paraphrase", "paraphrase"]))
```
