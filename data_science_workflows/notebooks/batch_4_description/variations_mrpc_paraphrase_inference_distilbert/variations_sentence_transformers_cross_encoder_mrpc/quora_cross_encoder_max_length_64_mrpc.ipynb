```python
def get_embeddings(text):
    return client.embeddings.create(
            input=text,
            model="text-embedding-3-large"
        ).data[0].embedding
```

```python
import torch
import numpy as np
from datasets import Dataset
from sentence_transformers import CrossEncoder
from sklearn.metrics import accuracy_score, f1_score, classification_report
from tqdm.auto import tqdm

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
print("device:", device)
```

```python
model_name = "cross-encoder/quora-distilroberta-base"
max_length = 64

model = CrossEncoder(model_name, device=str(device), max_length=max_length)
print("model:", model_name)
print("max_length:", max_length)
```

```python
ds = vault.query_item_content("glue_mrpc_validation")
ds = Dataset.from_list(ds)
print(ds)
print(ds[0])

sent1 = ds["sentence1"]
sent2 = ds["sentence2"]
y_true = np.array(ds["label"])
pairs = list(zip(sent1, sent2))

print("num_examples:", len(y_true))
print("positive_rate:", float(y_true.mean()))
```

```python
batch_size = 64

raw_outputs = model.predict(
    pairs,
    batch_size=batch_size,
    show_progress_bar=True,
    convert_to_numpy=True,
)

raw_outputs = np.asarray(raw_outputs)
print("raw_outputs.shape:", raw_outputs.shape)

if raw_outputs.ndim == 2 and raw_outputs.shape[1] == 2:
    logits = raw_outputs
    probs = torch.softmax(torch.tensor(logits), dim=-1).cpu().numpy()
    pos_scores = probs[:, 1]
    y_pred = np.argmax(logits, axis=-1).astype(int)
elif raw_outputs.ndim == 2 and raw_outputs.shape[1] == 1:
    pos_scores = raw_outputs[:, 0]
    y_pred = (pos_scores >= 0.5).astype(int)
elif raw_outputs.ndim == 1:
    pos_scores = raw_outputs
    y_pred = (pos_scores >= 0.5).astype(int)
else:
    raise ValueError(f"Unexpected prediction shape: {raw_outputs.shape}")

print("done")
```

```python

vault.create_record_list("distilbert_cross_text_prediction_cutoff", column_names=["prediction", "score"])

for i in range(len(y_pred)):
    vault.append_record("distilbert_cross_text_prediction_cutoff", 
                        {
                            "prediction": int(y_pred[i]),
                            "score": float(pos_scores[i]) ,
                        },
                       input_items = {
                           "glue_mrpc_validation": [i, i + 1],
                       }
                       )

description = "Per-example prediction outputs from the cross-encoder paraphrase model evaluated on the GLUE MRPC validation set. Each row corresponds to one sentence pair from glue_mrpc_validation and stores two fields: prediction (binary predicted label, where 1 indicates paraphrase and 0 indicates not_paraphrase) and score (the model\u2019s positive-class score for the pair; in this notebook, the softmax probability of the paraphrase class when logits are returned). This dataset is an intermediate inference artifact used to persist row-level model outputs, trace them back to the source MRPC examples, inspect correct and incorrect predictions, and support downstream summary evaluation such as accuracy, F1, and the classification report."
embedding = get_embeddings(description)
vault.create_description("distilbert_cross_text_prediction_cutoff", description, embedding)

properties = {"task": "paraphrase detection", "artifact_type": "model predictions", "prediction_type": "binary classification", "score_type": "positive class probability", "model": "cross-encoder/quora-distilroberta-base", "max_length": "64", "source": "glue/mrpc", "source_record_list": "glue_mrpc_validation", "split": "validation", "size": "408", "domain": "news", "input_fields": "sentence1,sentence2", "output_fields": "prediction,score", "label_space": "not_paraphrase,paraphrase"} #e.g. task: paraphrase detection

for prop, cat in properties.items():
    embedding = get_embeddings(prop)
    vault.create_description("distilbert_cross_text_prediction_cutoff", cat, embedding, prop)
```

```python
acc = accuracy_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
report = classification_report(y_true, y_pred, target_names=['not_paraphrase', 'paraphrase'])

print({"accuracy": acc, "f1": f1})
print(classification_report(y_true, y_pred, target_names=["not_paraphrase", "paraphrase"]))
```
